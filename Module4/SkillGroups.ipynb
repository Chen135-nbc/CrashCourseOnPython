```python
# 技能 1：使用字符串方法
def sales_quantity(order):
    items = []
    quantities = []
    # 使用.strip()去除多余的空白。
    order = order.strip()
    # 使用.split()将字符串中的数值与文本值分开。 
    sales_or_quantity = order.split()
    # 遍历字符串中的元素。
    for x in sales_or_quantity:
        # 使用.isalpha()测试元素是否包含字母.
        if x.isalpha():
            items.append(x)
            # 将分割后的字符串的各个元素分别赋值给新的变量。
        else:
            quantities.append(x)
    
    for i in range(len(items)):
        # 使用.format()和{ }变量占位符格式化字符串。
        print("{} has quantity {}".format(items[i],quantities[i]))
    
    

print(sales_quantity(" apple 3 banana 5 orange 2 "))
```

```python
# 技能1：使用字符串方法
def sales_prices(item_and_price):
    item = ""
    price = ""
    item_or_price = item_and_price.split()
    
    for x in item_or_price:
        if x.isalpha():
            item += x + " "
        else:
            price = x
    
    item = item.strip()
    return "{} are on sale for ${}".format(item, price)

print(sales_prices("Winter fleece jackets 49.99"))
```

```python
# 技能 2：使用列表方法
def food_order(order1,order2):
    # 使用.reverse()方法反转列表的顺序。
    order1.reverse()
    order2.reverse()
    # 使用.extend()方法合并两个列表。
    order2.extend(order1)
    return order2

order1 = ["apple", "banana", "orange"]
order2 = ["milk", "bread"]

print(food_order(order1,order2))
```

```python
# 技能 2：使用列表方法
def record_profit_years(recent_first, recent_last):
    recent_first.reverse()
    recent_last.extend(recent_first)
    return recent_last

recent_first = [2022, 2018, 2011, 2006]
recent_last = [1989, 1992, 1997, 2001]

print(record_profit_years(recent_first,recent_last))
```

```python
# 技能3：运用列表理解 
def calculate(lower,upper):
    # 使用列表推导式 [ ] 作为从范围创建新列表的快捷方式。
    # 使用for循环在具有 2 个参数（lower，upper+1）的范围内进行计算。
    # 使用列表推导式 [ ] 和for循环以及if条件语句
    num = [ x for x in range(lower,upper+1) if x % 3 ==0 ]
    result = [ i**2 for i in num]
    return result 

print(calculate(1,20))
```

```python
# 技能3：运用列表理解 
def calculate1(lower,upper):
    num1 = [ y for y in range(lower, upper+1) if y % 3 == 0 and y % 2 == 0]
    return [ i ** 2 for i in num1]
print(calculate1(1,30))
```
