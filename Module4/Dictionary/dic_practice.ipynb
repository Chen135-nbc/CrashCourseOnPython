```python
# 1. 创建一个dictionary，然后打印值
person = {
    "name": "Tom",
    "age": 20,
    "city": "London"
}
print(person["name"])
```

```python
# 2.给dictionary添加一个键值，打印整个dictionary
person = {
    "name": "Alice",
    "age": 25
}
person["city"] = "Berlin"
print(person)
```

```python
# 3.给定词典，用for循环打印
scores = {
    "Alice": 90,
    "Bob": 85,
    "Charlie": 92
}
for name, score in scores.items():
    print("{} scored {}".format(name,score))
```

```python
# 4.统计字符串中每个字母出现的次数
text = "apple"
dic = {}
for char in text:
    if char not in dic:
        dic[char] = 0
    dic[char] += 1
print(dic)

```

```python
# 统计字符串中每个字母出现的次数·简化写法
text = "apple"
dic = {}
for char in text:
    #如果 char 存在 → 返回 dic[char] ；如果不存在 → 返回 0
    dic[char] = dic.get(char,0) + 1
print(dic)
```

```python
# 5.统计一句话里 每个单词出现的次数
text = "apple banana apple orange banana apple"
words = text.split()
dic1 = {}
for word in words:
    if word not in dic1:
        dic1[word] = 0
    dic1[word] += 1
print(dic1)
```
