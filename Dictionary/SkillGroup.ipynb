```python
# Skill Group 1:
# 使用for循环和dictionary.items()方法遍历字典的键和值对，以计算字典中值的总和。
# 练习题：计算总分
scores = {
    "math": 90,
    "english": 85,
    "science": 88,
    "history": 92
}
total = 0
for key,value in scores.items():
    total += value
print(f"Total score: {total}")
    
```

```python
# Skill Group 2:
# 将字典中每个元素的值、字符串和键连接起来，并使用list.append(x)方法将其添加到新列表[ ] 的末尾。  
# 使用嵌套的for循环和dictionary.items()方法遍历字典中具有多个值的键。
students = {
    "Tom": ["math", "english"],
    "Alice": ["science", "math"],
    "Bob": ["history"]
}
result = []
for name, subjects in students.items():
    for subject in subjects:
        result.append(f"{name} studies {subject}")
print(result)
```

```python
# Skill Group 2:
products = {
    "apple": [5, 6],
    "banana": [3],
    "orange": [8, 9]
}
result1 = []

for product, prices in products.items():
    for price in prices:
        result1.append(f"{product} costs {price}")
print(result1)
```
