```python
# Practice 1: append
nums = []
for i in range(1,4):
    nums.append(i)
print(nums)
```

```python
# Practice 2: insert
nums1 = [1,3,4]
nums1.insert(1,2)
print(nums1)
```

```python
# Practice 3:extend
a = [1, 2]
b = [3, 4, 5]
a.extend(b)
print(a)
```

```python
# Practice 4:pop 按位置删 + 拿到值
nums2 = [10, 20, 30, 40]
x = nums2.pop(2)
print(x)
print(nums2)
```

```python
# Practice 5:pop()（默认删除最后一个）
nums3 = [1,2,3,4]
x1 = nums3.pop()
print(x1)
print(nums3)
```

```python
# Practice 6:remove（按值删）
nums4 = [1, 2, 2, 3, 4]
nums4.remove(2)
print(nums4)
```

```python
# Practice 7:sort()（最基础）
nums5 = [3, 1, 4, 2]
nums5.sort()
print(nums5)
```

```python
# Practice 8:reverse()（顺序反转）
nums6 = [1, 2, 3, 4, 5]
nums6.reverse()
print(nums6)
```

```python
# Practice 9: sort + reverse 组合
nums7 = [10, 3, 7, 1, 5]
nums7.sort()
nums7.reverse()
print(nums7)
```
